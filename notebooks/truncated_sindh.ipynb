```python
import pandas as pd

sindh = pd.read_csv("../data/processed/sindh.csv")
```

```python
with open("../data/processed/columns.txt", "w", encoding="utf-8") as f:
    for column in sindh.columns:
        f.write(f"{column}\n")
```

```python
with open("../data/processed/selected_columns.txt", "r", encoding="utf-8") as f:
    selected_columns = [line.strip() for line in f if line.strip()]
```

```python
selected_columns
```

```python
truncated_sindh = sindh[selected_columns].copy()
```

```python
truncated_sindh.shape
```

```python
truncated_sindh.head()
```

```python
truncated_sindh.to_csv(
    "../data/processed/truncated_sindh.csv",
    index=False
)
```

```python
for col in ["motive", "nperps"]:
    print(f"\n--- {col} ---")
    print("Null:", truncated_sindh[col].isna().sum())
    print("Null %:", truncated_sindh[col].isna().mean() * 100)
    print("Unique:", truncated_sindh[col].nunique())
```

```python
print("nperps = -99:", (truncated_sindh["nperps"] == -99).sum())
print("nperps = -99 %:", (truncated_sindh["nperps"] == -99).mean() * 100)
```
