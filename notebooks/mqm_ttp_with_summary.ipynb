```python
import pandas as pd
```

```python
sindh_mqm_ttp = pd.read_csv("../data/demo/sindh_mqm_ttp.csv")
```

```python
mqm_ttp_with_summary = sindh_mqm_ttp[
    sindh_mqm_ttp["summary"].notna() &
    sindh_mqm_ttp["summary"].str.strip().ne("")
].copy()
```

```python
mqm_ttp_with_summary.shape
```

```python
mqm_with_summary = mqm_ttp_with_summary[mqm_ttp_with_summary["gname"] == "Muttahida Qami Movement (MQM)"]
```

```python
mqm_with_summary.shape
```

```python
ttp_with_summary = mqm_ttp_with_summary[mqm_ttp_with_summary["gname"] == "Tehrik-i-Taliban Pakistan (TTP)"]
```

```python
ttp_with_summary.shape
```

```python
mqm_with_summary.to_csv("../data/demo/mqm_with_summary.csv")
```

```python
ttp_with_summary.to_csv("../data/demo/ttp_with_summary.csv")
```

```python
ttp_filtered = ttp_with_summary[
    (ttp_with_summary["iyear"].isin([2012, 2013])) &
    (ttp_with_summary["city"] == "Karachi") &
    (ttp_with_summary["summary"].notna()) &
    (ttp_with_summary["summary"].str.strip().ne(""))
].copy()

print("Candidate rows:", len(ttp_filtered))
```

```python
ttp_filtered.groupby(
    ["attacktype1_txt", "targtype1_txt"]
).size().sort_values(ascending=False)
```

```python
ttp_diverse = (
    ttp_filtered
    .sort_values("nkill", ascending=False)
    .groupby(
        ["attacktype1_txt", "targtype1_txt"],
        group_keys=False
    )
    .head(2)
)
```

```python
print("Diverse candidates:", len(ttp_diverse))

ttp_diverse[
    [
        "eventid",
        "iyear",
        "city",
        "attacktype1_txt",
        "weaptype1_txt",
        "targtype1_txt",
        "target1",
        "nkill",
        "nwound",
        "summary"
    ]
]
```

```python
print(ttp_diverse["attacktype1_txt"].value_counts())
print(ttp_diverse["targtype1_txt"].value_counts())
```
